# Model Explainability with SHAP

This notebook uses SHAP to explain model predictions and feature importance.

```python
import shap
import joblib
import pandas as pd
import matplotlib.pyplot as plt

print('✓ SHAP library loaded')
```

```python
# Load model
model = joblib.load('../models/admission_model.pkl')
scaler = joblib.load('../models/scaler.pkl')
feature_names = joblib.load('../models/feature_names.pkl')

print('✓ Model loaded')
```

```python
# Load test data
df = pd.read_csv('../data/processed/patient_data_processed.csv')
X = df.drop('admission', axis=1)
X['gender_encoded'] = (X['gender'] == 'M').astype(int)
X = X.drop('gender', axis=1)[feature_names]
X_scaled = scaler.transform(X)

print(f'Loaded {len(X)} samples')
```

```python
# Create SHAP explainer
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_scaled[:100])

print('✓ SHAP values calculated')
```

```python
# Summary plot
shap.summary_plot(shap_values[1], X_scaled[:100], feature_names=feature_names)
```

```python
# Feature importance
shap.summary_plot(shap_values[1], X_scaled[:100], feature_names=feature_names, plot_type='bar')
```
